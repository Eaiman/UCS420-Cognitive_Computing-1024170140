## Question 1


```python
import pandas as pd

roll_number_str = "1024170140"
categories = ["billing", "account", "general"]

fixed_entries = [
    {"question": "what is the annual fee", "answer": "The annual fee is Rs 500.", "keywords": "fee cost price charge", "category": "billing"},
    {"question": "how to reset password", "answer": "Go to Settings > Reset Password.", "keywords": "password reset login", "category": "account"},
    {"question": "what are your working hours", "answer": "We are open 9 AM to 5 PM.", "keywords": "hours timing open time", "category": "general"},
    {"question": "how can i pay the fee", "answer": "You can pay via UPI, card, or net banking.", "keywords": "pay payment upi fee", "category": "billing"}
]
```

```python
if not roll_number_str.isdigit() or len(roll_number_str) < 2:
    raise ValueError("Enter a valid roll number.")

first_digit, second_digit = map(int, roll_number_str[-2:])
dynamic_entries = []

def make_entry(digit, kind):
    category = categories[digit % len(categories)]
    if kind == 1:
        templates = {
            "billing": (f"When is payment {digit} due?", f"Payment {digit} is due on the {digit}th of each month.", f"payment due date {digit}"),
            "account": (f"How do I update my email for account {digit}?", f"Update the email from account settings for item {digit}.", f"email update account {digit}"),
            "general": (f"Where are service details for {digit}?", f"Service details for {digit} are in the website help section.", f"service details help {digit}")
        }
    else:
        templates = {
            "billing": (f"How do I request a refund for item {digit}?", f"Contact support to request a refund for item {digit}.", f"refund support transaction {digit}"),
            "account": (f"What if I forget username code {digit}?", f"Use the Forgot Username option for account item {digit}.", f"username forgot account {digit}"),
            "general": (f"What features were added for {digit}?", f"Check announcements for updates related to {digit}.", f"features updates announcements {digit}")
        }
    question, answer, keywords = templates[category]
    return {"question": question, "answer": answer, "keywords": keywords, "category": category}

dynamic_entries = [make_entry(first_digit, 1), make_entry(second_digit, 2)]
for item in dynamic_entries:
    print(item)
```

```python
all_entries = fixed_entries + dynamic_entries
df_faq = pd.DataFrame(all_entries)
display(df_faq)
```

## Question 2

```python
import re

def score_faq_entries(query, df):
    query_words = set(re.findall(r"[a-z0-9]+", query.lower()))
    scores = []

    for _, row in df.iterrows():
        entry_text = " ".join(str(row[field]) for field in ["question", "answer", "keywords"]).lower()
        entry_words = set(re.findall(r"[a-z0-9]+", entry_text))
        scores.append(len(query_words & entry_words))

    results = df.copy()
    results["confidence_score"] = scores
    results = results[results["confidence_score"] > 0]
    return results.sort_values("confidence_score", ascending=False).reset_index(drop=True)

for query in ["how to pay my fees", "reset my password please", "information about 8 services"]:
    print("Search query:", query)
    display(score_faq_entries(query, df_faq))
```

## Question 3

```python
def same_category(category_name, df):
    selected = df.loc[df["category"].str.casefold() == category_name.casefold(), "question"]
    return selected.tolist()

chosen_category = dynamic_entries[0]["category"]
for number, question in enumerate(same_category(chosen_category, df_faq), start=1):
    print(f"{number}. {question}")
```

## Question 4

```python
entry_idx = 0
print("Before update:")
display(df_faq.iloc[entry_idx])

keyword = input("Enter a keyword to add: ").strip() or "discount"
current = df_faq.at[entry_idx, "keywords"]
words = current.split()
if keyword.lower() not in [word.lower() for word in words]:
    words.append(keyword)
    df_faq.at[entry_idx, "keywords"] = " ".join(words)

print("After update:")
display(df_faq.iloc[entry_idx])
csv_filename = f"{roll_number_str}_faq_data.csv"
df_faq.to_csv(csv_filename, index=False)
print("Saved file:", csv_filename)
```

## Question 6

```python
def score_faq_entries_v2(query, df):
    ranked = score_faq_entries(query, df)
    if ranked.empty:
        print("No matching FAQ entries found.")
        return ranked

    best_score = ranked["confidence_score"].iloc[0]
    best_matches = ranked[ranked["confidence_score"] == best_score]
    if len(best_matches) > 1:
        print("More than one entry shares the best score:", best_score)
        display(best_matches)
    else:
        print("Best match score:", best_score)
        display(best_matches)
    return ranked

for search_text in ["how to pay fee", "reset password"]:
    print("Query:", search_text)
    score_faq_entries_v2(search_text, df_faq)
```
