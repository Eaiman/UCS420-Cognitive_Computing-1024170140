# Question 2


```python
roll = input("Enter Roll No.: ").strip()
digit_values = [int(ch) * 10 for ch in roll[-8:][::-1]]
scores = tuple(digit_values)
print("Scores:", scores)

if scores:
    top_score = max(scores)
    bottom_score = min(scores)
    print("Highest score:", top_score)
    print("Index of highest score:", scores.index(top_score))
    print("Lowest score:", bottom_score)
    print("Lowest score frequency:", scores.count(bottom_score))
    print("Reversed tuple as list:", list(reversed(scores)))

score_to_find = int(input("Enter a score: "))
print("First occurrence index:", scores.index(score_to_find) if score_to_find in scores else "Score not present")
```

# Question 3

```python
import random

random.seed(1024170140)
values = [random.randrange(100, 901) for _ in range(100)]
odd_values = [v for v in values if v % 2]
even_values = [v for v in values if v % 2 == 0]

print("Odd numbers:", odd_values)
print("Count of odd numbers:", len(odd_values))
print("Even numbers:", even_values)
print("Count of even numbers:", len(even_values))

def prime_check(value):
    if value < 2:
        return False
    for divisor in range(2, int(value ** 0.5) + 1):
        if value % divisor == 0:
            return False
    return True

prime_values = [v for v in values if prime_check(v)]
print("Prime numbers:", prime_values)
print("Count of prime numbers:", len(prime_values))

frequency = {v: values.count(v) for v in set(values)}
most_common = max(frequency, key=frequency.get)
print("Most frequent number:", most_common)
print("Frequency:", frequency[most_common])
```

# Question 4

```python
roll = input("Enter Roll No.: ").strip()
digits = [int(char) for char in roll[:8] if char.isdigit()]
set_a = {digit * 7 for digit in digits}
set_b = {digit * 9 for digit in digits}

print("A:", set_a)
print("B:", set_b)
print("Union:", set_a | set_b)
print("Intersection:", set_a & set_b)
print("A - B:", set_a - set_b)
print("B - A:", set_b - set_a)
print("Symmetric difference:", set_a ^ set_b)
print("A is subset of B:", set_a <= set_b)
print("B is superset of A:", set_b >= set_a)

remove_value = int(input("Enter a value to remove from A: "))
set_a.discard(remove_value)
print("A after discard:", set_a)
```
