```python
import numpy as np

temperature = np.array([25, 28, 31, 35, 38, 27, 33, 40])
corrected = np.add(temperature, 2)
fahrenheit = temperature * 1.8 + 32
hot_readings = temperature[temperature > 32]

print("a. Corrected Temperatures (Celsius):", corrected)
print("b. Temperatures in Fahrenheit:", fahrenheit)
print("c. Readings greater than 32 degrees C:", hot_readings)
print("d. Count of readings exceeding 32 degrees C:", hot_readings.size)
print("\ne. Vectorization applies one operation to an entire NumPy array instead of writing a Python loop. NumPy performs these operations efficiently in compiled code. Boolean indexing uses a True/False condition to select only the values that meet a requirement.")
```

### Question 2


```python
import numpy as np

steps = np.array([
    [5000, 6200, 7100],
    [8000, 7500, 9000],
    [4500, 5100, 4800],
    [9000, 8500, 9500]
])

print("a. Total steps recorded:", steps.sum())
print("b. Mean number of steps:", steps.mean())
print("c. Maximum steps:", steps.max(), "Minimum steps:", steps.min())
print("d. Total steps for each day:", steps.sum(axis=0))
print("e. Total steps for each user:", steps.sum(axis=1))
best_flat_index = int(steps.argmax())
best_position = np.unravel_index(best_flat_index, steps.shape)
print("f. Flat argmax index:", best_flat_index, "Grid position (User, Day):", best_position)
```

### Question 4


```python
import numpy as np

X = np.array([[6, 70, 3], [5, 50, 6], [8, 80, 2], [4, 30, 8]])
y = np.array([40, 65, 30, 85])

print("a. Shape of X:", X.shape, "Dimensions:", X.ndim)
transpose = np.transpose(X)
print("b. Transpose of X:\n", transpose)
print("Transpose swaps the rows and columns.")

product = transpose @ X
print("c. X.T @ X:\n", product)
inverse_product = np.linalg.inv(product)
print("d. Inverse of X.T @ X:\n", inverse_product)
coefficients = inverse_product @ transpose @ y
print("e. Calculated coefficients:", coefficients)

print("\nf. Coefficient meaning:")
print(f"Sleep-hours coefficient ({coefficients[0]:.2f}) estimates score change for one extra sleep hour.")
print(f"Activity coefficient ({coefficients[1]:.2f}) estimates score change for one activity unit.")
print(f"Stress coefficient ({coefficients[2]:.2f}) estimates score change for one stress point.")

new_user = np.array([5, 40, 7])
print("g. Predicted score for user [5, 40, 7]:", round(float(new_user @ coefficients), 2))
```
