### Question 2: Reversing array and finding most frequent values

```python
arr_to_reverse = np.array([1, 2, 3, 6, 4, 5])
print("a) Reversed array:", np.flip(arr_to_reverse))

x = np.array([1, 2, 3, 4, 5, 1, 2, 1, 1, 1])
y = np.array([1, 1, 1, 2, 3, 4, 2, 4, 3, 3])

def frequent_indices(data):
    values, counts = np.unique(data, return_counts=True)
    highest = counts.max()
    answer = {}
    for value, count in zip(values, counts):
        if count == highest:
            answer[int(value)] = np.flatnonzero(data == value)
    return answer

print("b) i. For array x:", frequent_indices(x))
print("b) ii. For array y:", frequent_indices(y))
```

### Question 5: Array manipulations, statistical measures, reshaping and resizing

```python
data_2d = np.array([[10, 20, 30, 40], [50, 60, 70, 80], [90, 15, 20, 35]])
print("Original 2D Array:\n", data_2d)
print("Mean:", data_2d.mean())
print("Median:", np.median(data_2d))
print("Max:", data_2d.max())
print("Min:", data_2d.min())
print("Unique elements:", np.unique(data_2d))
print("Reshaped Array (4x3):\n", data_2d.reshape(4, 3))
print("Resized Array (2x3):\n", np.resize(data_2d, (2, 3)))
```
